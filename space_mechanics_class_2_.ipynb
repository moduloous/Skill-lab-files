```python
import pandas as pd
import statsmodels.formula.api as sm

# Load your dataset
air = pd.read_csv("https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/main/New_Atm_Para.csv")

# 1. Model 1: Altitude vs Density

model1 = sm.ols(formula='Density ~ Altitude', data=air)
fitted1 = model1.fit()
print("\nRegression Summary: Altitude vs Density")
print(fitted1.summary())
```

```python
import pandas as pd

# Create a new data point for prediction
new_data = pd.DataFrame({'Altitude': [9872]})

# Use your previously fitted model (fitted1)
predicted_density = fitted1.predict(new_data)

print("\nPredicted Density at 9872 m:")
print(predicted_density)
```

```python
import pandas as pd
import statsmodels.formula.api as smf

# Load dataset
air = pd.read_csv("https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/main/New_Atm_Para.csv")

# Define models
models = {
    "Altitude": 'Density ~ Altitude',
    "Temperature": 'Density ~ Temperature',
    "Pressure": 'Density ~ Pressure',
    "Wind_Gust": 'Density ~ Wind_Gust'
}

fitted_models = {}

# Fit and display summaries
for name, formula in models.items():
    model = smf.ols(formula=formula, data=air).fit()
    fitted_models[name] = model

    print(f"\nRegression Summary: Density vs {name}")
    print(model.summary())

# ISA Values for Prediction
isa_values = {
    "Altitude": 8900,
    "Temperature": 230.3,
    "Pressure": 31195.25,
    "Wind_Gust": 11.43
}

# Prediction & R²
for name, value in isa_values.items():
    new_data = pd.DataFrame({name: [value]})
    prediction = fitted_models[name].predict(new_data)[0]

    print(f"\nPredicted Density using {name}: {prediction:.4f} kg/m³")
    print(f"R² ({name} Model): {fitted_models[name].rsquared:.4f}")
```

```python
import pandas as pd
import statsmodels.formula.api as smf

# Load dataset (URL must be inside quotes)
air = pd.read_csv(
    "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/main/New_Atm_Para.csv"
)
# Multiple Linear Regression Model
model = smf.ols(
    formula='Density ~ Altitude + Temperature + Pressure + Mach_Number + Relative_Humidity + Wind_Gust + Turbulence_Intensity',
    data=air
)
fitted = model.fit()
print("\nMultiple Regression Summary:")
print(fitted.summary())
```
